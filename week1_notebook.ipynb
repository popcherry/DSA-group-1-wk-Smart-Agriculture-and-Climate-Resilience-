### 1.2 A list
A list holds several items in order. This one holds the crops we study.

```python
crops = ["Maize", "Wheat", "Rice", "Potatoes", "Soybeans",
         "Sorghum", "Cassava", "Sweet potatoes", "Barley", "Millet"]

print("Number of crops:", len(crops))
print("First crop:", crops[0])
print("Last crop:", crops[-1])
```

### 1.3 A dictionary
A dictionary links a key to a value. Here each crop is linked to its crop group.

```python
crop_group = {
    "Maize": "Cereal",
    "Wheat": "Cereal",
    "Rice": "Cereal",
    "Barley": "Cereal",
    "Sorghum": "Cereal",
    "Millet": "Cereal",
    "Potatoes": "Root/Tuber",
    "Cassava": "Root/Tuber",
    "Sweet potatoes": "Root/Tuber",
    "Soybeans": "Legume",
}

print("Maize belongs to:", crop_group["Maize"])
```

### 1.4 A loop
A loop repeats an action for every item. We print each crop with its group.

```python
for crop in crops:
    print(crop, "->", crop_group[crop])
```

### 1.6 Two small functions
A function packages code we can reuse.

- `yield_per_ha` calculates yield in kg per hectare.
- `classify_rainfall` puts annual rainfall into a simple band.

```python
def yield_per_ha(yield_kg, area_ha):
    """Return yield in kg per hectare. Returns None if area is zero or negative."""
    if area_ha <= 0:
        return None
    return yield_kg / area_ha


def classify_rainfall(rainfall_mm):
    """Return a rainfall band for annual rainfall in millimetres."""
    if rainfall_mm < 500:
        return "Dry"
    elif rainfall_mm < 1200:
        return "Moderate"
    else:
        return "Wet"


# quick tests
print(yield_per_ha(5000000, 2000))      # 2500.0 kg/ha
print(classify_rainfall(350))            # Dry
print(classify_rainfall(900))            # Moderate
print(classify_rainfall(2000))           # Wet
```
