### 1.1 Variables
Variables store single values we reuse in the project.

### 1.2 A list
A list holds several items in order. This one holds the 10 crops in our dataset.

```python
crops = [
    "Maize (corn)", "Potatoes", "Wheat", "Rice", "Sweet potatoes",
    "Sorghum", "Cassava, fresh", "Soya beans", "Yams", "Plantains and cooking bananas"
]

print("Number of crops:", len(crops))
print("List of crops:", crops)
```

### 1.3 A dictionary
A dictionary links a key to a value. Here each crop is linked to its crop group.

```python
crop_group = {
    "Maize (corn)": "Cereal",
    "Wheat": "Cereal",
    "Rice": "Cereal",
    "Sorghum": "Cereal",
    "Potatoes": "Root/Tuber",
    "Sweet potatoes": "Root/Tuber",
    "Cassava, fresh": "Root/Tuber",
    "Yams": "Root/Tuber",
    "Soya beans": "Legume",
    "Plantains and cooking bananas": "Fruit/Starchy Staple"
}
```

### 1.4 A loop
A loop repeats an action for every item. We print each crop with its group.

```python
for crop in crops:
    print(crop, "->", crop_group[crop])
```

### 1.5 A conditional
An `if` statement lets the code make a decision. We use it to check the project rules.

```python
no_of_rows = 25432

if no_of_rows >= min_rows_required:
    print("Row rule met:", no_of_rows)
else:
    print("Row rule NOT met:", no_of_rows)

no_of_columns = 14

if no_of_columns >= min_columns_required:
    print("Column rule met:", no_of_columns)
else:
    print("Column rule NOT met:", no_of_columns)
```

### 1.6 Two small functions
A function packages code we can reuse.

- `yield_per_ha` calculates yield in kg per hectare from total tonnes and harvested area.
- `classify_rainfall` puts annual rainfall into a simple band.

```python
def yield_per_ha(production_tonnes, area_ha):
    """Return yield in kg per hectare. Returns None if area is zero or negative."""
    if area_ha <= 0:
        return None
    return (production_tonnes * 1000) / area_ha


def classify_rainfall(rainfall_mm):
    """Return a rainfall band for annual rainfall in millimetres."""
    if rainfall_mm < 500:
        return "Dry"
    elif rainfall_mm < 1200:
        return "Moderate"
    else:
        return "Wet"


# Test both functions with sample values (no manual input() needed)
sample_rainfall = 1400
sample_production_tonnes = 320
sample_area_ha = 200

print("Rainfall band:", classify_rainfall(sample_rainfall))
print("Yield per hectare:", yield_per_ha(sample_production_tonnes, sample_area_ha), "kg/ha")
```

## Part 2: Load the dataset with Pandas and prove the size rules

```python
import pandas as pd
from pathlib import Path

data_path = Path("../data/raw/agri_merged_raw.csv")
df = pd.read_csv(data_path)
print("Dataset loaded successfully!")
```

### 2.1 Shape, columns and first rows

```python
print("rows, columns:", df.shape)
print("column names:", df.columns)
print("\nfirst 5 rows:")
df.head()
```

### 2.2 Check the rules using the conditional

```python
no_of_rows, no_of_cols = df.shape

if no_of_rows >= min_rows_required:
    print(f"Rows: {no_of_rows} (>= {min_rows_required}) PASS")
else:
    print(f"Rows: {no_of_rows} (< {min_rows_required}) FAIL")

if no_of_cols >= min_columns_required:
    print(f"Columns: {no_of_cols} (>= {min_columns_required}) PASS")
else:
    print(f"Columns: {no_of_cols} (< {min_columns_required}) FAIL")

print("Duplicate rows:", df.duplicated().sum())
```

### 2.3 Use the functions on the real data

```python
# Using a complete record from the dataset (e.g., last row: Zimbabwe Wheat 2013)
row = df.iloc[-1]

rainfall = row["rainfall_mm"]
production_t = row["production_tonnes"]
area_ha = row["plot_area_ha"]

rainfall_band = classify_rainfall(rainfall)
yield_result = yield_per_ha(production_t, area_ha)

print(f"Record: {row['country']} | {row['crop']} | {row['year']}")
print("Rainfall band:", rainfall_band)
print("Calculated yield per hectare:", round(yield_result, 2), "kg/ha")
print("Recorded yield_kg_ha:", row["yield_kg_ha"], "kg/ha")
```
